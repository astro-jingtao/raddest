```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ks_2samp, norm
```

```python
x = np.random.normal(0, 1, 1000)
x = np.append(x, [100])
```

```python
norm.fit(x)
```

```python
def gen_sample(mu, sigma, n):
    return np.random.normal(mu, sigma, n)

def target(par, x, n):
    mu, sigma = par
    x_gen = gen_sample(mu, sigma, n)
    return ks_2samp(x, x_gen).statistic
```

```python
# optimization
from scipy.optimize import minimize


minimize(target, [0, 3], args=(x, 1000), method='Nelder-Mead').x
```
